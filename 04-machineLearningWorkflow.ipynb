# Machine Learning Workflow

A Machine Learning workflow is the sequence of steps followed to build, evaluate, and use a machine learning model.

```
Problem Definition
       ↓
Data Collection
       ↓
Data Understanding
       ↓
Data Preprocessing
       ↓
Feature Engineering
       ↓
Train-Test Split
       ↓
Model Selection
       ↓
Model Training
       ↓
Model Evaluation
       ↓
Model Improvement
       ↓
Deployment
       ↓
Monitoring
```

## Step 1 — Problem Definition

What problem are we trying to solve?

For example:

A bank wants to predict whether a transaction is fraudulent.

Here:

Input → Transaction information

Output → Fraud / Not Fraud

We need to determine:

- What are we predicting?
- What data do we have?
- Is this regression or classification?
- What does success mean?

```
Why this step matters

If you don't clearly define the problem, you might build a technically good model that solves the wrong problem.
```
