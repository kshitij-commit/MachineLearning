# Machine Learning Workflow

A Machine Learning workflow is the sequence of steps followed to build, evaluate, and use a machine learning model.

```
Problem Definition
       ↓
Data Collection
       ↓
Data Understanding
       ↓
Data Preprocessing
       ↓
Feature Engineering
       ↓
Train-Test Split
       ↓
Model Selection
       ↓
Model Training
       ↓
Model Evaluation
       ↓
Model Improvement
       ↓
Deployment
       ↓
Monitoring
```

## Step 1 — Problem Definition

What problem are we trying to solve?

For example:

A bank wants to predict whether a transaction is fraudulent.

Here:

Input → Transaction information

Output → Fraud / Not Fraud

We need to determine:

- What are we predicting?
- What data do we have?
- Is this regression or classification?
- What does success mean?

```
Why this step matters

If you don't clearly define the problem, you might build a technically good model that solves the wrong problem.
```

## Step 2 — Data Collection

Now we need data.

For the fraud example, we might collect:

- Transaction amount
- Transaction time
- Location
- Payment method
- Customer information
- Previous transaction history
- Fraud label

Data can come from:

- Databases
- APIs
- Sensors
- Websites
- Files
- Company systems
- Surveys

The quality of the data has a huge effect on the quality of the final model.

## Step 3 — Data Understanding

We ask questions such as:

- How many rows are there?
- What features exist?
- Which columns contain missing values?
- Are there duplicate records?
- What are the data types?
- Are there unusual values?
- How is the target distributed?

```
For example:

Dataset
├── Age
├── Income
├── Transaction Amount
├── Location
└── Fraud

We shouldn't immediately train a model.

First:

Understand the data.
```
