# Machine Learning Workflow

A Machine Learning workflow is the sequence of steps followed to build, evaluate, and use a machine learning model.

```
Problem Definition
       ↓
Data Collection
       ↓
Data Understanding
       ↓
Data Preprocessing
       ↓
Feature Engineering
       ↓
Train-Test Split
       ↓
Model Selection
       ↓
Model Training
       ↓
Model Evaluation
       ↓
Model Improvement
       ↓
Deployment
       ↓
Monitoring
```
