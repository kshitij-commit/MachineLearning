## 1. Supervised Learning

The model learns from data where the correct answer (label/target) is already provided.

```
Example:

Hours Studied	Exam Score
2	            45
4	            60
6	            72
8	            85
```
Here:

- Input/features (X) → Hours Studied
- Output/target (y) → Exam Score

The model learns:

Hours studied → Score

Then we can give:
```
10 hours
   ↓
Model
   ↓
? score
```

More Example-
- house price prediction
- spam detection

2.Classification

### Regression vs Classification

1. Regression :- Regression is a supervised learning technique used when the output we want to predict is a continuous numerical value.

Examples:

- Predict house price → ₹45 lakh
- Predict temperature → 32.5°C
- Predict salary → ₹6.8 lakh/year
- Predict exam score → 78.5

2. classification :- Classification is a supervised learning technique used when the output belongs to a category/class.

Examples:

- Email → Spam / Not Spam
- Tumor → Benign / Malignant
- Transaction → Fraud / Not Fraud
- Image → Cat / Dog


## 2. Unsupervised Learning

````python
The model receives data without predefined labels and tries to discover hidden patterns or structures.
````

Example :- customer segmentation

Customer segmentation means dividing customers into different groups (segments) based on similarities in their characteristics or behavior.

```
Example:

Suppose an online store has:

Customer	Spending	Visits
A	            ₹5,000	20
B	            ₹4,500	18
C	            ₹500	3
D	            ₹700	4
```
We don't tell the model which customers belong together.

The algorithm might discover:

Group 1 → High-value customers
A, B

Group 2 → Low-value customers
C, D

Common unsupervised techniques

Association

- Apriori

## 3. Semi-Supervised Learning

````python
The model learns from a small amount of labeled data and a large amount of unlabeled data.
````

```
Example:

Suppose you have 100,000 images.

Only:

1,000 → labeled
99,000 → unlabeled

Manually labeling 100,000 images would be expensive.

Instead:

Small labeled dataset
        +
Large unlabeled dataset
        ↓
Semi-supervised learning
        ↓
Better model
```

Example applications
- Medical image classification
- Speech recognition
- Web content classification
- Image recognition

## 4.Reinforcement Learning

An agent learns by interacting with an environment and receiving rewards or penalties for its actions.

```
Think about teaching a dog.

Dog performs action
       ↓
Correct action?
       ↓
Reward

Over time, the dog learns which actions produce rewards.

Similarly:

Agent
  ↓
Action
  ↓
Environment
  ↓
Reward / Penalty
  ↓
Agent learns
```
