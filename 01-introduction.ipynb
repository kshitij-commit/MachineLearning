# B. Machine Learning (ML)

Machine Learning is a way of building systems that learn patterns from data and use those patterns to make predictions or decisions.

```
Instead of manually writing every rule:

IF email contains "lottery"
AND email contains "winner"
AND email contains "prize"
THEN spam

we can provide examples:

Email                         Label

"Win $1000!"                  Spam
"Meeting at 3 PM"             Not Spam
"Congratulations winner"      Spam
"Project report attached"     Not Spam
```

# C. Deep Learning (DL)

Deep Learning is a subset of Machine Learning that uses multi-layer neural networks to learn complex patterns from data.

```
For example, suppose we want to recognize cats.

Traditional ML might require us to manually create features such as:

Ear shape
Eye shape
Body shape
Fur texture

Deep learning can learn useful representations automatically from raw images.

Image
  ↓
Neural Network
  ↓
Edges
  ↓
Shapes
  ↓
Parts
  ↓
Object
  ↓
"Cat"
```

# D. Data Science

Data Science is the process of extracting useful information, insights, and predictions from data using statistics, programming, data analysis, visualization, and machine learning.

Suppose Amazon has millions of orders.

A data scientist might ask:

- Which products sell the most?
- Which customers are leaving?
- What factors affect sales?
- Which products should we recommend?
- How will sales change next month

```
Raw Data
   ↓
Collect
   ↓
Clean
   ↓
Explore
   ↓
Analyze
   ↓
Visualize
   ↓
Build ML Model (if needed)
   ↓
Insights / Predictions
   ↓
Business Decision
```
