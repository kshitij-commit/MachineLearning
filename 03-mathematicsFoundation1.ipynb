# ML Mathematics Foundation — Part 1

## 1. What is an Equation of a Line?



$$ y = mx + c $$

This describes a straight line.

For example:

$$ y = 2x + 3 $$

Let's understand what every part means.

- x → input
- y → output
- m → slope
- c → y-intercept

So:

$$ \boxed{y = mx + c} $$

means:

Start at c, then change y by m for every 1-unit change in x.
	​


### 2. What is Slope?

Slope tells us how quickly y changes when x changes.

Mathematically:

$$ m = \frac{\Delta y}{\Delta x} $$

where:

- Δx = change in x
- Δy = change in y

Example

Suppose:

- x = 1 → y = 3
- x = 2 → y = 5

When x increases by 1:

$$ \Delta x = 2-1 = 1 $$

y increases by:

$$ \Delta y = 5-3 = 2 $$

Therefore:

$$ m = \frac{2}{1}=2 $$

So the slope is 2.

That means:

Whenever x increases by 1, y increases by 2.

## 3. Understanding the Intercept

Consider:

$$ y = 2x + 3 $$

The 3 is the y-intercept.

It tells us the value of y when:

$$ x=0 $$

Put \(x=0\):

$$ y=2(0)+3 $$ $$ y=3 $$

So the line crosses the y-axis at:

(0,3)

## 4. Let's Calculate Some Points

Take:

$$ y=2x+3 $$

If:

x = 0
$$ y=2(0)+3=3 $$
x = 1
$$ y=2(1)+3=5 $$
x = 2
$$ y=2(2)+3=7 $$
x = 3
$$ y=2(3)+3=9 $$

So:
```
x	y
0	3
1	5
2	7
3	9

Notice the pattern:

x increases by 1
        ↓
y increases by 2
```
That's exactly what the slope = 2 tells us.
