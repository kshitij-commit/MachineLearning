# ML Mathematics Foundation — Part 1

## 1. What is an Equation of a Line?



$$ y = mx + c $$

This describes a straight line.

For example:

$$ y = 2x + 3 $$

Let's understand what every part means.

- x → input
- y → output
- m → slope
- c → y-intercept

So:

$$ \boxed{y = mx + c} $$

means:

Start at c, then change y by m for every 1-unit change in x.
	​


### 2. What is Slope?

Slope tells us how quickly y changes when x changes.

Mathematically:

$$ m = \frac{\Delta y}{\Delta x} $$

where:

- Δx = change in x
- Δy = change in y

Example

Suppose:

- x = 1 → y = 3
- x = 2 → y = 5

When x increases by 1:

$$ \Delta x = 2-1 = 1 $$

y increases by:

$$ \Delta y = 5-3 = 2 $$

Therefore:

$$ m = \frac{2}{1}=2 $$

So the slope is 2.

That means:

Whenever x increases by 1, y increases by 2.
